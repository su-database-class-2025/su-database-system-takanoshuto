東京都の市区町村ごとの2019年4月の休日昼間の人口密度を地図化する

```python
import os
from sqlalchemy import create_engine
import pandas as pd
import geopandas as gpd
import numpy as np
import folium
pd.set_option('display.max_columns', 100)
```

```python
def query_geopandas(sql, db):
    """
    Executes a SQL query on a postGIS and returns the result as a GeoPandas GeoDataFrame.

    Args:
        sql (str): The SQL query to execute.
        db (str): The name of the PostgreSQL database to connect to.

    Returns:
        geopandas.GeoDataFrame: The result of the SQL query as a GeoPandas GeoDataFrame.
    """
    DATABASE_URL = 'postgresql://postgres:postgres@postgis_container:5432/{}'.format(db)
    conn = create_engine(DATABASE_URL)
    query_result_gdf = gpd.GeoDataFrame.from_postgis(
        sql, conn, geom_col='geom') #geom_col='way' when using osm_kanto, geom_col='geom' when using gisdb
    return query_result_gdf
```

## Define a sql command

```python

sql = "with pop as \
            (select distinct(p.name), d.prefcode, d.year, d.month, d.population, p.geom \
                from pop as d \
                    inner join pop_mesh as p \
                        on p.name = d.mesh1kmid \
                    where d.dayflag='0' and \
                            d.timezone='0' and \
                            d.year='2019' and \
                            d.month='04') \
        select poly.name_2, pop.population as sum, pop.year, pop.month, pop.prefcode, poly.geom \
            from pop \
                inner join adm2 as poly \
                    on st_within(pop.geom, poly.geom) \
            where poly.name_1='Tokyo' \
            group by poly.name_2, pop.year, pop.month, pop.prefcode, poly.geom \
            order by pop.population desc;"
```

## Outputs

```python
def display_interactive_map(gdf):
    m = folium.Map(location=[36, 139.5], zoom_start=10)

    folium.Choropleth(
        geo_data=gdf.to_json(),
        data=gdf,
        columns=['name_2', 'sum'],
        key_on='feature.properties.name_2',
        fill_color='YlOrRd',
        fill_opacity=0.7,
        line_opacity=0.2,
        bins = [1, 10000, 100000, 300000, 500000, 700000, 1000000, 1200000]
    ).add_to(m)

    return m
```

```python
out = query_geopandas(sql,'gisdb')
map_display = display_interactive_map(out)
print(out)
display(map_display)
```
